# Git Tutorial - Part 2: Viewing History & Changes
## Detailed Scenarios and Use Cases

## 4. Viewing Commit History

### 4.1 Basic Log Commands

**Why view history?**
- Understand what changed and when
- Find when bugs were introduced
- Review team member contributions
- Generate changelogs

```python
# View full commit history
# SCENARIO: You want to see all commits in the project
# WHY: Shows complete history with full details
# OUTPUT: Commit hash, author, date, message
# WHEN: Investigating project history, writing documentation
!git log

# One-line format (most commonly used)
# SCENARIO: Quick overview of recent changes
# WHY: Compact, easy to scan
# OUTPUT: Short hash + commit message
# WHEN: Daily workflow, quick checks
!git log --oneline

# Limit number of commits
# SCENARIO: You only care about recent changes
# WHY: Avoids overwhelming output
!git log -n 10  # Last 10 commits
!git log -5     # Last 5 commits
```

```python
# Visual graph of branches
# SCENARIO: Complex branching, need to understand merge history
# WHY: Shows branch relationships visually
# WHEN: Debugging merge issues, understanding workflow
!git log --graph --oneline --all

# Detailed graph with dates
# SCENARIO: Need to see when branches diverged
!git log --graph --oneline --all --decorate --date=relative

# Pretty format (customized)
# SCENARIO: Creating reports or documentation
# WHY: Full control over output format
!git log --pretty=format:"%h - %an, %ar : %s"
# %h = short hash
# %an = author name
# %ar = author date, relative
# %s = subject (commit message)
```

### 4.2 Filtering Commits

**Real-world scenarios for filtering:**

```python
# Filter by author
# SCENARIO: Reviewing a specific developer's work
# WHY: See what one person contributed
# WHEN: Code reviews, performance reviews, debugging
!git log --author="John Doe"
!git log --author="john"  # Partial match works

# Filter by date range
# SCENARIO: Need commits from last sprint/month
# WHY: Time-based analysis, changelog generation
!git log --since="2 weeks ago"
!git log --since="2024-01-01"
!git log --after="2024-01-01" --before="2024-01-31"
!git log --since="1 month ago" --until="2 days ago"

# Search commit messages
# SCENARIO: Finding when a feature was added
# WHY: Locate specific changes by keyword
# WHEN: Bug hunting, feature tracking
!git log --grep="authentication"
!git log --grep="fix" --grep="bug"  # OR condition
!git log --grep="fix" --grep="bug" --all-match  # AND condition
```

```python
# Show commits that modified specific file
# SCENARIO: Who changed this file and when?
# WHY: Track file evolution, find bug introduction
# WHEN: Debugging file-specific issues
!git log -- path/to/file.py
!git log --follow -- file.py  # Follows renames

# Show actual changes (patches)
# SCENARIO: Need to see what actually changed in each commit
# WHY: Detailed code review
# CAUTION: Very verbose output
!git log -p
!git log -p -2  # Last 2 commits with changes

# Show statistics
# SCENARIO: How much code was changed?
# WHY: Quick overview of change magnitude
!git log --stat
!git log --shortstat  # Just numbers
```

### 4.3 Git Blame - Line-by-Line History

**When to use blame:**
- Finding who wrote specific code
- Understanding why code was written
- Not for blaming, but for understanding context!

```python
# Show who last modified each line
# SCENARIO: You found a bug and need to know who wrote it
# WHY: Shows author, date, and commit for each line
# WHEN: Debugging, understanding code history
!git blame filename.py

# Blame specific lines
# SCENARIO: Bug is on lines 50-60
# WHY: Focus on relevant section
!git blame -L 50,60 filename.py

# Ignore whitespace changes
# SCENARIO: File was reformatted, hiding real changes
# WHY: See actual code changes, not formatting
!git blame -w filename.py

# Show email addresses
# SCENARIO: Need to contact the author
!git blame -e filename.py
```

## 5. Viewing Changes (Diff)

### 5.1 Understanding Diff Output

**Diff shows:**
- Lines added (green, +)
- Lines removed (red, -)
- Context lines (white)

**When to use diff:**
- Before committing (review changes)
- Comparing branches
- Code review

```python
# Show unstaged changes
# SCENARIO: You modified files, want to see what changed
# WHY: Review before staging
# WHEN: Before 'git add', multiple times per day
# SHOWS: Working directory vs. staging area
!git diff

# Show staged changes
# SCENARIO: You staged files, want to review before commit
# WHY: Final check before committing
# WHEN: After 'git add', before 'git commit'
# SHOWS: Staging area vs. last commit
!git diff --staged
!git diff --cached  # Same thing

# Show all changes (staged + unstaged)
# SCENARIO: Want to see everything that changed
!git diff HEAD
```

```python
# Diff specific file
# SCENARIO: Only care about changes in one file
# WHY: Focus on relevant changes
!git diff filename.py
!git diff --staged filename.py

# Word-level diff (not line-level)
# SCENARIO: Small changes in long lines
# WHY: Highlights exact words that changed
# WHEN: Documentation, config files
!git diff --word-diff

# Show statistics only
# SCENARIO: How many files/lines changed?
# WHY: Quick overview without details
!git diff --stat
!git diff --numstat  # Machine-readable format
```

### 5.2 Comparing Branches and Commits

```python
# Compare two branches
# SCENARIO: What's different between main and feature branch?
# WHY: Before merging, understand changes
# WHEN: Code review, pre-merge checks
!git diff main..feature-branch
!git diff main...feature-branch  # Since branches diverged

# Compare two commits
# SCENARIO: What changed between two versions?
# WHY: Understand evolution between releases
!git diff abc123 def456
!git diff HEAD~2 HEAD  # Last 2 commits

# Compare with remote branch
# SCENARIO: What will be pushed/pulled?
# WHY: Preview before push/pull
!git diff main origin/main

# Show only file names that changed
# SCENARIO: Just want to know which files, not details
!git diff --name-only main..feature
!git diff --name-status main..feature  # With status (M/A/D)
```

### 5.3 Show Command - Inspect Commits

```python
# Show specific commit details
# SCENARIO: Want to see what a commit changed
# WHY: Full details: message, author, date, changes
# WHEN: Investigating specific commits
!git show abc1234

# Show last commit
# SCENARIO: What did I just commit?
# WHY: Quick verification
!git show HEAD
!git show HEAD~1  # Second to last
!git show HEAD~2  # Third to last

# Show specific file from commit
# SCENARIO: How did this file look in that commit?
# WHY: View file content at specific point
!git show abc1234:path/to/file.py

# Show commit without diff
# SCENARIO: Just want commit info, not changes
!git show --no-patch abc1234
!git show -s abc1234  # Same thing
```
