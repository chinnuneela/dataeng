# Comprehensive Git Tutorial
## From Basics to Advanced Concepts

This notebook covers everything you need to know about Git - from basic commands to advanced workflows and troubleshooting.

## 1. Git Basics

### 1.1 What is Git?
Git is a distributed version control system that tracks changes in source code during software development.

### 1.2 Installation & Configuration

````python
# Check Git version
!git --version

# Configure user information
!git config --global user.name "Your Name"
!git config --global user.email "your.email@example.com"

# View all configurations
!git config --list

# Set default editor
!git config --global core.editor "vim"

# Set default branch name
!git config --global init.defaultBranch main
````

### 1.3 Creating a Repository

````python
# Initialize a new Git repository
!git init

# Initialize with specific branch name
!git init -b main

# Clone an existing repository
!git clone https://github.com/username/repository.git

# Clone to a specific directory
!git clone https://github.com/username/repository.git my-folder

# Clone a specific branch
!git clone -b branch-name https://github.com/username/repository.git
````

## 2. Basic Git Workflow

### 2.1 Staging and Committing

````python
# Check repository status
!git status

# Add files to staging area
!git add filename.txt
!git add .
!git add -A
!git add *.py

# Add interactively
!git add -p

# Commit changes
!git commit -m "Commit message"

# Commit with detailed message
!git commit -m "Title" -m "Detailed description"

# Add and commit in one step
!git commit -am "Message"

# Amend last commit
!git commit --amend -m "Updated message"

# Amend without changing message
!git commit --amend --no-edit
````

### 2.2 Viewing History

````python
# View commit history
!git log

# One line per commit
!git log --oneline

# With graph
!git log --graph --oneline --all

# Last N commits
!git log -n 5

# Show changes in commits
!git log -p

# Filter by author
!git log --author="John"

# Filter by date
!git log --since="2 weeks ago"
!git log --after="2024-01-01" --before="2024-12-31"

# Search commit messages
!git log --grep="bug fix"

# Show specific file history
!git log -- filename.txt

# Show who changed what
!git blame filename.txt
````

### 2.3 Viewing Changes

````python
# Show unstaged changes
!git diff

# Show staged changes
!git diff --staged
!git diff --cached

# Compare branches
!git diff branch1..branch2

# Compare commits
!git diff commit1 commit2

# Show changes in specific file
!git diff filename.txt

# Show statistics
!git diff --stat

# Show specific commit
!git show commit-hash
!git show HEAD
````

## 3. Branching and Merging

### 3.1 Branch Management

````python
# List branches
!git branch
!git branch -a  # Include remote branches
!git branch -v  # With last commit

# Create new branch
!git branch feature-branch

# Switch to branch
!git checkout feature-branch
!git switch feature-branch  # Modern command

# Create and switch in one step
!git checkout -b feature-branch
!git switch -c feature-branch

# Rename branch
!git branch -m old-name new-name

# Delete branch
!git branch -d feature-branch
!git branch -D feature-branch  # Force delete

# Delete remote branch
!git push origin --delete feature-branch
````

### 3.2 Merging

````python
# Merge branch into current branch
!git merge feature-branch

# Merge with commit message
!git merge feature-branch -m "Merge message"

# No fast-forward merge (creates merge commit)
!git merge --no-ff feature-branch

# Fast-forward only
!git merge --ff-only feature-branch

# Squash merge (combine all commits)
!git merge --squash feature-branch

# Abort merge
!git merge --abort
````

## 4. Remote Repositories

### 4.1 Remote Management

````python
# List remotes
!git remote
!git remote -v  # With URLs

# Add remote
!git remote add origin https://github.com/username/repo.git

# Change remote URL
!git remote set-url origin https://github.com/username/new-repo.git

# Remove remote
!git remote remove origin

# Rename remote
!git remote rename old-name new-name

# Show remote info
!git remote show origin
````

### 4.2 Fetching and Pulling

````python
# Fetch from remote (doesn't merge)
!git fetch origin
!git fetch --all

# Fetch and prune deleted branches
!git fetch --prune

# Pull (fetch + merge)
!git pull origin main
!git pull

# Pull with rebase
!git pull --rebase origin main

# Pull all branches
!git pull --all
````

### 4.3 Pushing

````python
# Push to remote
!git push origin main

# Push and set upstream
!git push -u origin main

# Push all branches
!git push --all

# Push tags
!git push --tags

# Force push (dangerous!)
!git push --force
!git push --force-with-lease  # Safer option

# Push new branch
!git push -u origin feature-branch
````

## 5. Undoing Changes

### 5.1 Unstaging and Discarding

````python
# Unstage file
!git reset HEAD filename.txt
!git restore --staged filename.txt  # Modern command

# Discard changes in working directory
!git checkout -- filename.txt
!git restore filename.txt  # Modern command

# Discard all changes
!git checkout .
!git restore .

# Remove untracked files
!git clean -n  # Dry run
!git clean -f  # Force
!git clean -fd  # Include directories
!git clean -fdx  # Include ignored files
````

### 5.2 Reset

````python
# Soft reset (keep changes staged)
!git reset --soft HEAD~1

# Mixed reset (keep changes unstaged) - default
!git reset HEAD~1
!git reset --mixed HEAD~1

# Hard reset (discard all changes)
!git reset --hard HEAD~1

# Reset to specific commit
!git reset --hard commit-hash

# Reset to remote state
!git reset --hard origin/main
````

### 5.3 Revert

````python
# Revert a commit (creates new commit)
!git revert commit-hash

# Revert without committing
!git revert -n commit-hash

# Revert multiple commits
!git revert commit1..commit2

# Revert merge commit
!git revert -m 1 merge-commit-hash
````

## 6. Advanced Git Concepts

### 6.1 Rebasing

````python
# Rebase current branch onto main
!git rebase main

# Interactive rebase (last 3 commits)
!git rebase -i HEAD~3

# Continue rebase after resolving conflicts
!git rebase --continue

# Skip current commit
!git rebase --skip

# Abort rebase
!git rebase --abort

# Rebase onto specific commit
!git rebase --onto new-base old-base branch
````

### 6.2 Cherry-picking

````python
# Apply specific commit to current branch
!git cherry-pick commit-hash

# Cherry-pick multiple commits
!git cherry-pick commit1 commit2

# Cherry-pick range
!git cherry-pick commit1..commit2

# Cherry-pick without committing
!git cherry-pick -n commit-hash

# Continue after resolving conflicts
!git cherry-pick --continue

# Abort cherry-pick
!git cherry-pick --abort
````

### 6.3 Stashing

````python
# Stash changes
!git stash
!git stash save "Work in progress"

# Stash including untracked files
!git stash -u

# Stash including ignored files
!git stash -a

# List stashes
!git stash list

# Show stash contents
!git stash show
!git stash show -p

# Apply stash
!git stash apply
!git stash apply stash@{1}

# Apply and remove stash
!git stash pop

# Drop stash
!git stash drop stash@{0}

# Clear all stashes
!git stash clear

# Create branch from stash
!git stash branch new-branch
````

### 6.4 Tags

````python
# List tags
!git tag
!git tag -l "v1.*"

# Create lightweight tag
!git tag v1.0.0

# Create annotated tag
!git tag -a v1.0.0 -m "Version 1.0.0"

# Tag specific commit
!git tag -a v1.0.0 commit-hash -m "Message"

# Show tag info
!git show v1.0.0

# Delete tag
!git tag -d v1.0.0

# Delete remote tag
!git push origin --delete v1.0.0

# Push tag
!git push origin v1.0.0

# Push all tags
!git push --tags
````

### 6.5 Submodules

````python
# Add submodule
!git submodule add https://github.com/user/repo.git path/to/submodule

# Initialize submodules
!git submodule init

# Update submodules
!git submodule update

# Clone with submodules
!git clone --recursive https://github.com/user/repo.git

# Update all submodules
!git submodule update --remote

# Remove submodule
!git submodule deinit path/to/submodule
!git rm path/to/submodule
````

### 6.6 Reflog

````python
# Show reflog
!git reflog

# Show reflog for specific branch
!git reflog show branch-name

# Recover lost commit
!git reset --hard HEAD@{1}

# Create branch from reflog entry
!git branch recovered-branch HEAD@{2}
````

## 7. Conflict Resolution

### 7.1 Understanding Merge Conflicts

Conflicts occur when:
- Same line modified in different branches
- File deleted in one branch but modified in another
- Binary files modified differently

Conflict markers:
```
<<<<<<< HEAD
Your changes
=======
Their changes
>>>>>>> branch-name
```

### 7.2 Resolving Conflicts

````python
# Check conflict status
!git status

# See conflicted files
!git diff --name-only --diff-filter=U

# View conflict in detail
!git diff

# Accept their version
!git checkout --theirs filename.txt

# Accept our version
!git checkout --ours filename.txt

# After manual resolution, mark as resolved
!git add filename.txt

# Continue merge
!git commit

# Abort merge
!git merge --abort
````

### 7.3 Merge Tools

````python
# Configure merge tool
!git config --global merge.tool vimdiff

# Launch merge tool
!git mergetool

# Available tools: vimdiff, meld, kdiff3, etc.
!git mergetool --tool=meld
````

### 7.4 Rebase Conflicts

````python
# During rebase, resolve conflicts then:
!git add resolved-file.txt
!git rebase --continue

# Skip problematic commit
!git rebase --skip

# Abort rebase
!git rebase --abort
````

## 8. Git Workflows

### 8.1 Feature Branch Workflow

````python
# Create feature branch
!git checkout -b feature/new-feature

# Work on feature
# ... make changes ...
!git add .
!git commit -m "Add new feature"

# Push feature branch
!git push -u origin feature/new-feature

# After review, merge to main
!git checkout main
!git merge feature/new-feature
!git push origin main

# Delete feature branch
!git branch -d feature/new-feature
!git push origin --delete feature/new-feature
````

### 8.2 Gitflow Workflow

````python
# Main branches: main (production), develop (integration)

# Start feature
!git checkout -b feature/my-feature develop

# Finish feature
!git checkout develop
!git merge --no-ff feature/my-feature
!git branch -d feature/my-feature

# Start release
!git checkout -b release/1.0.0 develop

# Finish release
!git checkout main
!git merge --no-ff release/1.0.0
!git tag -a v1.0.0
!git checkout develop
!git merge --no-ff release/1.0.0
!git branch -d release/1.0.0

# Hotfix
!git checkout -b hotfix/1.0.1 main
# ... fix bug ...
!git checkout main
!git merge --no-ff hotfix/1.0.1
!git tag -a v1.0.1
!git checkout develop
!git merge --no-ff hotfix/1.0.1
!git branch -d hotfix/1.0.1
````

## 9. Advanced Troubleshooting

### 9.1 Common Issues and Solutions

````python
# Issue: Detached HEAD state
# Solution: Create branch or return to branch
!git checkout -b new-branch  # Create branch from current state
!git checkout main  # Return to main branch

# Issue: Accidentally committed to wrong branch
# Solution: Move commits to correct branch
!git log  # Note commit hash
!git checkout correct-branch
!git cherry-pick commit-hash
!git checkout wrong-branch
!git reset --hard HEAD~1

# Issue: Need to undo last commit but keep changes
!git reset --soft HEAD~1

# Issue: Committed sensitive data
!git filter-branch --force --index-filter \
  "git rm --cached --ignore-unmatch path/to/file" \
  --prune-empty --tag-name-filter cat -- --all

# Better: Use BFG Repo-Cleaner
# bfg --delete-files sensitive.txt

# Issue: Large file causing push to fail
!git rm --cached large-file.bin
!git commit --amend -C HEAD
````

### 9.2 Recovering Lost Work

````python
# Find lost commits
!git reflog

# Recover deleted branch
!git reflog
!git checkout -b recovered-branch HEAD@{n}

# Find dangling commits
!git fsck --lost-found

# Recover from hard reset
!git reflog
!git reset --hard HEAD@{n}
````

### 9.3 Fixing Merge Conflicts - Advanced

````python
# Rerere (Reuse Recorded Resolution)
!git config --global rerere.enabled true

# This remembers how you resolved conflicts
# and automatically applies same resolution next time

# View three-way diff during conflict
!git show :1:filename  # Common ancestor
!git show :2:filename  # Our version
!git show :3:filename  # Their version

# Resolve using strategy
!git merge -X ours branch-name  # Prefer our changes
!git merge -X theirs branch-name  # Prefer their changes
````

## 17. Best Practices

### 17.1 Commit Messages

Good commit message format:
```
Short summary (50 chars or less)

More detailed explanation if needed. Wrap at 72 characters.
Explain what and why, not how.

- Bullet points are okay
- Use imperative mood: "Add feature" not "Added feature"

Fixes #123
```

### 17.2 Branching Strategy

- Use descriptive branch names: `feature/user-auth`, `bugfix/login-error`
- Keep branches short-lived
- Delete merged branches
- Regularly sync with main branch

### 17.3 General Tips

- Commit often, push regularly
- Write meaningful commit messages
- Review changes before committing
- Use `.gitignore` properly
- Never commit sensitive data
- Keep commits atomic (one logical change)
- Test before pushing
- Use pull requests for code review
