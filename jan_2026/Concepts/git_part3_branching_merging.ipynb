## 6. Understanding Branches

### What is a Branch?
- A **lightweight movable pointer** to a commit
- Allows **parallel development** without affecting main code
- **Cheap to create** (just a 41-byte file!)

### Why Use Branches?
1. **Feature Development**: Work on new features without breaking main code
2. **Bug Fixes**: Fix bugs in isolation
3. **Experimentation**: Try ideas without commitment
4. **Collaboration**: Multiple developers work simultaneously
5. **Release Management**: Maintain multiple versions

### Common Branch Names:
- `main`/`master`: Production-ready code
- `develop`: Integration branch
- `feature/feature-name`: New features
- `bugfix/bug-description`: Bug fixes
- `hotfix/critical-fix`: Urgent production fixes
- `release/v1.0.0`: Release preparation

## 7. Branch Management

### 7.1 Listing Branches

````python
# List local branches
# SCENARIO: Want to see all branches in your repo
# WHY: Know what branches exist
# OUTPUT: List with * marking current branch
!git branch

# List with last commit info
# SCENARIO: Need context about each branch
# WHY: See what each branch is about
# WHEN: Deciding which branch to work on
!git branch -v

# List all branches (including remote)
# SCENARIO: Want to see remote branches too
# WHY: Know what's on the server
# WHEN: Looking for teammates' branches
!git branch -a

# List remote branches only
# SCENARIO: What branches exist on GitHub/GitLab?
!git branch -r

# List merged branches
# SCENARIO: Which branches can be safely deleted?
# WHY: Cleanup old branches
# WHEN: Maintenance, keeping repo clean
!git branch --merged

# List unmerged branches
# SCENARIO: Which branches still have unique work?
# WHY: Avoid deleting important work
!git branch --no-merged
````

### 7.2 Creating Branches

````python
# Create new branch (doesn't switch to it)
# SCENARIO: Want to create branch but stay on current one
# WHY: Prepare branch for later
# WHEN: Planning work, creating multiple branches
!git branch feature/user-authentication

# Create branch from specific commit
# SCENARIO: Need to branch from an older commit
# WHY: Start work from specific point in history
# WHEN: Creating hotfix from production version
!git branch hotfix/security-patch abc1234

# Create branch from tag
# SCENARIO: Need to patch old release
# WHY: Branch from exact release point
!git branch bugfix/v1.0-patch v1.0.0
````

### 7.3 Switching Branches

````python
# Switch to existing branch (old way)
# SCENARIO: Move to different branch
# WHY: Work on different feature/fix
# WHAT HAPPENS: Updates working directory to match branch
!git checkout feature/user-auth

# Switch to branch (modern way - Git 2.23+)
# SCENARIO: Same as above
# WHY: Clearer command (switch vs checkout)
# RECOMMENDED: Use this instead of checkout
!git switch feature/user-auth

# Create and switch in one command (old way)
# SCENARIO: Start new feature immediately
# WHY: Most common workflow
# WHEN: Beginning new work
!git checkout -b feature/new-dashboard

# Create and switch (modern way)
# SCENARIO: Same as above
# RECOMMENDED: Clearer syntax
!git switch -c feature/new-dashboard

# Switch to previous branch
# SCENARIO: Quickly toggle between two branches
# WHY: Convenience, like 'cd -'
# WHEN: Comparing changes, frequent switching
!git switch -
!git checkout -  # Old way
````

**⚠️ Important: Uncommitted Changes**

When switching branches:
- Git prevents switching if changes would be lost
- Options:
  1. Commit changes
  2. Stash changes (covered later)
  3. Discard changes (careful!)

### 7.4 Renaming Branches

````python
# Rename current branch
# SCENARIO: Typo in branch name or better name found
# WHY: Keep naming consistent
# WHEN: Before pushing to remote
!git branch -m new-branch-name

# Rename different branch
# SCENARIO: Rename branch you're not on
!git branch -m old-name new-name

# Rename and update remote
# SCENARIO: Already pushed old name to remote
# WHY: Keep local and remote in sync
# STEPS:
!git branch -m old-name new-name  # Rename locally
!git push origin :old-name  # Delete old remote branch
!git push origin new-name  # Push new branch
!git push origin -u new-name  # Set upstream
````

### 7.5 Deleting Branches

````python
# Delete merged branch (safe)
# SCENARIO: Feature merged to main, cleanup time
# WHY: Keep repo clean, remove clutter
# WHEN: After successful merge
# SAFETY: Git prevents deleting if not merged
!git branch -d feature/completed-feature

# Force delete branch (dangerous!)
# SCENARIO: Abandoned feature, want to discard work
# WHY: Remove unmerged branch
# CAUTION: Loses all unique commits!
# WHEN: Absolutely sure you don't need the work
!git branch -D feature/abandoned-idea

# Delete remote branch
# SCENARIO: Cleanup remote after merge
# WHY: Keep remote repo clean
# WHEN: After pull request merged
!git push origin --delete feature/old-feature
# OR
!git push origin :feature/old-feature  # Old syntax

# Delete multiple branches
# SCENARIO: Cleanup many old branches
!git branch -d branch1 branch2 branch3
````

## 8. Merging Branches

### What is Merging?
- **Combining** changes from different branches
- **Integrating** feature work back into main branch
- Creates a **merge commit** (usually)

### Types of Merges:
1. **Fast-Forward**: No divergent changes, just moves pointer
2. **Three-Way Merge**: Divergent changes, creates merge commit
3. **Squash Merge**: Combines all commits into one

### 8.1 Basic Merging

````python
# Standard merge workflow
# SCENARIO: Feature complete, merge to main
# STEPS:
# 1. Switch to target branch (main)
!git switch main
# 2. Merge feature branch
!git merge feature/user-auth

# WHY THIS ORDER: You merge FROM feature INTO main
# WHAT HAPPENS: 
# - If fast-forward possible: Just moves main pointer
# - If diverged: Creates merge commit

# Merge with custom message
# SCENARIO: Want descriptive merge commit
# WHY: Better history documentation
!git merge feature/payment -m "Merge payment feature - adds Stripe integration"
````

### 8.2 Fast-Forward vs No-Fast-Forward

````python
# No fast-forward merge (creates merge commit always)
# SCENARIO: Want to preserve feature branch history
# WHY: Shows feature as a unit in history
# WHEN: Team policy, important features
# RESULT: Always creates merge commit, even if could fast-forward
!git merge --no-ff feature/important-feature

# Fast-forward only (fails if not possible)
# SCENARIO: Want clean linear history
# WHY: Avoid merge commits when possible
# WHEN: Simple updates, rebased branches
# RESULT: Fails if branches diverged
!git merge --ff-only feature/simple-update
````

**Visual Example:**

Fast-Forward:
```
Before:  main: A---B
         feature:    C---D

After:   main: A---B---C---D
```

No-Fast-Forward:
```
Before:  main: A---B
         feature:    C---D

After:   main: A---B-------M
                     \     /
                      C---D
```

### 8.3 Squash Merging

````python
# Squash merge (combine all commits into one)
# SCENARIO: Feature has messy commit history
# WHY: Clean up history, one commit per feature
# WHEN: Many WIP commits, want clean main branch
# RESULT: All feature commits become one commit
# NOTE: Doesn't create merge commit automatically!
!git merge --squash feature/messy-history
!git commit -m "Add complete user dashboard feature"

# USE CASE:
# Feature branch has: "WIP", "fix typo", "oops", "actually fix"
# Main gets: "Add user dashboard with all features"
````

### 8.4 Aborting Merges

````python
# Abort merge (return to pre-merge state)
# SCENARIO: Merge conflicts too complex, need different approach
# WHY: Start over, try different strategy
# WHEN: During conflict resolution, before completing merge
# RESULT: Returns to state before merge command
!git merge --abort

# REAL-WORLD SCENARIO:
# 1. Start merge: git merge feature
# 2. See 50 conflicts
# 3. Realize you should rebase instead
# 4. Abort: git merge --abort
# 5. Try different approach
````
