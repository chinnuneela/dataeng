## 1. Git Basics & Configuration

### 1.1 What is Git?

**Git** is a distributed version control system that:
- Tracks changes in your code over time
- Allows multiple developers to work together
- Enables you to revert to previous versions
- Works offline (distributed nature)

**Key Concepts:**
- **Repository (Repo)**: A project folder tracked by Git
- **Commit**: A snapshot of your code at a specific point
- **Branch**: An independent line of development
- **Remote**: A version of your repo hosted elsewhere (GitHub, GitLab, etc.)

### 1.2 Initial Configuration

**Why configure Git?**
- Git needs to know who you are for commit attribution
- Configuration affects how Git behaves
- Settings can be global (all repos) or local (specific repo)

**When to configure:**
- First time using Git on a new machine
- When starting a new job (different email)
- When customizing Git behavior

````python
# Check Git version - ensures Git is installed
!git --version

# SCENARIO: You just installed Git and need to set up your identity
# WHY: Every commit you make will be tagged with this information
# WHEN: First time setup, or when changing identity
!git config --global user.name "John Doe"
!git config --global user.email "john.doe@company.com"

# View all configurations
# WHY: To verify settings or troubleshoot issues
!git config --list

# View specific config
!git config user.name
!git config user.email
````

````python
# Set default editor for commit messages
# SCENARIO: You prefer VSCode over vim for writing commit messages
# WHY: Makes writing detailed commit messages easier
!git config --global core.editor "code --wait"
# Other options: "vim", "nano", "emacs", "subl -n -w"

# Set default branch name to 'main' instead of 'master'
# SCENARIO: Following modern naming conventions
# WHY: Industry standard is shifting from 'master' to 'main'
# WHEN: Setting up Git for the first time
!git config --global init.defaultBranch main

# Enable colored output
# WHY: Makes Git output easier to read
!git config --global color.ui auto
````

**Configuration Levels:**
- `--system`: Applies to all users on the system
- `--global`: Applies to all repos for current user
- `--local`: Applies only to current repo (default)

## 2. Creating & Cloning Repositories

### 2.1 Initializing a New Repository

**Scenario:** You're starting a new project from scratch

````python
# Initialize Git in current directory
# SCENARIO: You have a project folder and want to start tracking it
# WHY: Converts a regular folder into a Git repository
# WHAT IT DOES: Creates a hidden .git folder with Git metadata
!git init

# Initialize with specific branch name
# SCENARIO: You want to use 'main' instead of default 'master'
# WHY: Sets the initial branch name explicitly
# WHEN: Creating new repos with modern naming
!git init -b main

# Initialize in a new directory
# SCENARIO: Creating a new project from scratch
!git init my-new-project
# This creates the directory AND initializes Git in it
````

### 2.2 Cloning Existing Repositories

**Scenario:** You want to work on an existing project

````python
# Clone a repository
# SCENARIO: You want to contribute to an open-source project
# WHY: Downloads the entire project history to your machine
# WHAT IT DOES: Creates a folder, downloads all files and commit history
!git clone https://github.com/username/repository.git

# Clone to a specific directory name
# SCENARIO: The repo name is too long or unclear
# WHY: Gives you control over the local folder name
!git clone https://github.com/username/very-long-repository-name.git myapp

# Clone a specific branch
# SCENARIO: You only need the 'develop' branch, not 'main'
# WHY: Saves time and bandwidth by only downloading one branch
# WHEN: Working on a specific feature or version
!git clone -b develop https://github.com/username/repository.git

# Shallow clone (limited history)
# SCENARIO: Large repo, you only need recent commits
# WHY: Faster download, less disk space
# WHEN: CI/CD pipelines, quick testing
!git clone --depth 1 https://github.com/username/repository.git
````

## 3. Basic Git Workflow

### The Three States of Git:
1. **Working Directory**: Where you edit files
2. **Staging Area (Index)**: Files ready to be committed
3. **Repository (.git directory)**: Committed snapshots

### 3.1 Checking Status

**Most used command** - Always check status before committing!

````python
# Check repository status
# SCENARIO: You've made changes and want to see what's modified
# WHY: Shows which files are modified, staged, or untracked
# WHEN: Before every commit, multiple times per day
# OUTPUT: Lists modified, staged, and untracked files
!git status

# Short status (concise output)
# SCENARIO: You want a quick overview
# WHY: Cleaner output, easier to scan
!git status -s
# Output format:
# M  = Modified (staged)
#  M = Modified (not staged)
# A  = Added (new file, staged)
# ?? = Untracked
````

### 3.2 Adding Files (Staging)

**Why stage files?**
- Allows you to commit only specific changes
- Lets you review changes before committing
- Enables logical grouping of changes

````python
# Add a specific file
# SCENARIO: You modified app.py and want to commit only that file
# WHY: Precise control over what gets committed
# WHEN: You've changed multiple files but want to commit them separately
!git add app.py

# Add multiple specific files
# SCENARIO: You fixed bugs in two related files
!git add file1.py file2.py file3.js

# Add all files in current directory
# SCENARIO: You've made changes across many files for one feature
# WHY: Faster than adding files individually
# CAUTION: Make sure you want to stage everything!
!git add .

# Add all files in repository (including deletions)
# SCENARIO: You've added, modified, and deleted files
# WHY: Stages all changes including deleted files
!git add -A
# OR
!git add --all

# Add all files with specific extension
# SCENARIO: You modified all Python files
# WHY: Pattern matching for selective staging
!git add *.py
!git add src/*.js
````

````python
# Interactive staging (ADVANCED)
# SCENARIO: You made multiple changes in one file but want to commit them separately
# WHY: Allows you to stage specific parts (hunks) of a file
# WHEN: You mixed bug fixes with new features in same file
# HOW IT WORKS: Git shows each change and asks if you want to stage it
!git add -p filename.py
# Options during interactive add:
# y = stage this hunk
# n = don't stage this hunk
# s = split into smaller hunks
# e = manually edit the hunk
# q = quit
````

### 3.3 Committing Changes

**What is a commit?**
- A snapshot of your staged changes
- Has a unique ID (SHA hash)
- Contains author, date, and message
- Permanent record in Git history

````python
# Basic commit with message
# SCENARIO: You've staged files and are ready to save the snapshot
# WHY: Creates a permanent record of your changes
# WHEN: After staging related changes
!git commit -m "Add user authentication feature"

# Commit with detailed message
# SCENARIO: Complex change that needs explanation
# WHY: First line is summary, second is detailed description
# BEST PRACTICE: Use for non-trivial changes
!git commit -m "Fix database connection timeout" -m "Increased timeout from 5s to 30s to handle slow networks. Added retry logic with exponential backoff."

# Add and commit in one step (tracked files only)
# SCENARIO: Quick fix to already tracked files
# WHY: Skips the staging step for convenience
# CAUTION: Only works for files already in Git, not new files
!git commit -am "Fix typo in README"
````

````python
# Amend the last commit
# SCENARIO 1: You forgot to include a file in your last commit
# SCENARIO 2: You made a typo in the commit message
# WHY: Modifies the most recent commit instead of creating a new one
# WHEN: Immediately after committing, before pushing
# CAUTION: Don't amend commits that have been pushed!

# Forgot a file? Stage it and amend:
!git add forgotten_file.py
!git commit --amend --no-edit  # Keeps same message

# Fix commit message:
!git commit --amend -m "Correct commit message"

# Open editor to modify message:
!git commit --amend
````

**Good Commit Message Guidelines:**
```
Short summary (50 chars max) - imperative mood

Detailed explanation if needed (wrap at 72 chars):
- What changed
- Why it changed
- Any side effects or considerations

Fixes #123
```

**Examples:**
- ✅ Good: "Add password reset functionality"
- ❌ Bad: "Added stuff"
- ✅ Good: "Fix null pointer exception in user service"
- ❌ Bad: "bug fix"
