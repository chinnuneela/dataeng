```python
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
import pyspark.sql.functions as F

# Initialize Spark session
spark = SparkSession.builder \
    .appName("Manager") \
    .getOrCreate()

data = [[101, 'John', 'A', None], [102, 'Dan', 'A', 101], [103, 'James', 'A', 101], [104, 'Amy', 'A', 101], [105, 'Anne', 'A', 101], [106, 'Ron', 'B', 101]]
employee = pd.DataFrame(data, columns=['id', 'name', 'department', 'managerId']).astype({'id':'Int64', 'name':'object', 'department':'object', 'managerId':'Int64'})

employee_spark = spark.createDataFrame(employee)
print("Employee DataFrame :")
employee_spark.show()

joined_df = employee_spark.alias('e1')\
    .join(employee_spark.alias('e2'), col('e1.managerId') == col('e2.id'), how='left')\
    .select(col('e1.id').alias('employee_id'), col('e1.name').alias('employee_name'),\
            col('e1.department').alias('department'), col('e2.name').alias('manager_name')).groupBy('manager_name')\
.agg(F.count('manager_name').alias('manager_count'))\
.filter(col('manager_count') >= 5)\
.select('manager_name', 'manager_count')\
.orderBy('manager_name')

print("Joined DataFrame :")
joined_df.show()

manager_count_df = 


spark.stop()
```
