```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, expr, round
import os


spark = SparkSession.builder \
    .appName("MacLargeDataGen") \
    .config("spark.driver.memory", "4g") \
    .getOrCreate()

 
num_rows = 10_00

# Generate the data
df = spark.range(0, num_rows) \
    .withColumn("transaction_id", col("id")) \
    .withColumn("user_id", (expr("rand()") * 100000).cast("int")) \
    .withColumn("category", expr("case when rand() > 0.5 then 'Electronics' else null end")) \
    .withColumn("timestamp", expr("to_timestamp('2023-01-01') + (rand() * 31536000) * interval 1 second")) \
    .withColumn("name", expr("case when rand() > 0.5 then 'Alice' else 'BOB' end")) \
    .withColumn("email", expr("case when rand() > 0.5 then 'user@example.com' else null end")) \
    .withColumn("amount", expr("case when rand() > 0.5 then rand() * 500 else null end")) \
    .withColumn("score", expr("case when rand() > 0.5 then rand() * 100 else null end")) \
    .withColumn("category_id", expr("case when rand() > 0.5 then CAST(rand() * 10 AS INT) else null end")) \
    .withColumn("discount", expr("case when rand() > 0.5 then rand() * 50 else null end")) \
    .withColumn("date_mm_dd_yyyy", expr("case when rand() > 0.5 then date_format(timestamp, 'MM-dd-yyyy') else null end")) \
    .withColumn("date_dd_mm_yyyy", expr("case when rand() > 0.5 then date_format(timestamp, 'dd-MM-yyyy') else null end")) \
    .withColumn("date_yyyy_mm_dd", expr("case when rand() > 0.5 then date_format(timestamp, 'yyyy-MM-dd') else null end"))

 

output_path = "large_practice_data.parquet"
df.write.mode("overwrite").parquet(output_path)

size_bytes = sum(os.path.getsize(os.path.join(dirpath, f)) 
                 for dirpath, dirnames, filenames in os.walk(output_path) 
                 for f in filenames)

print(f"✅ Success! File generated at: {os.path.abspath(output_path)}")
print(f"📊 Total Rows: {df.count():,}")
print(f"💾 Physical Size: {size_bytes / (1024**3):.2f} GB")

spark.stop()
```
