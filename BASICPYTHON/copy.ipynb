**Shallow Copy Vs Deep Copy**
 

```python
import copy 
```

```python
# Need some research on this ... 

# ls -ltr 
list1 = [1,2,[3,4]]

#list [1,2,3,4,[]]

s_copy = copy.copy(list1)


print(f"list1 : {list1} \n s_copy : {s_copy}")


s_copy[2][1] = 100 
s_copy[1] = 500

print(f"list1 : {list1} \n s_copy : {s_copy}")


```

```python
list1 = [1,2,[3,4]]

d_copy = copy.deepcopy(list1)

print(f"list1 : {list1} \n s_copy : {d_copy}")

d_copy[2][1] = 100 

print(f"list1 : {list1} \n s_copy : {s_copy}")
```
