```
Modes : 
r - read mode 
w - write ( create new file if not there , else overwrites )
a - append 
x - create ( fail if file already exists)
b - binary mode 
t - text mode 
```

````python
# Read from a file 

file_pointer = open(PATH,"r")
content = file_pointer.read()

print(content)

file_pointer.close()
````

````python
# Read from a file 

file_pointer = open(PATH,"r")
content = file_pointer.readlines()

print(content)

file_pointer.close()
````

````python
file2 = open("yash.txt","w")
file2.write("my content")
file2.close()
````

````python
with open("yash.txt","r") as file2:
    content = file2.read()
    print(content)

print(content)    
````
