## Basics 1
### Create some list in python 
```
1. Insert elements into the list 
2. Delete some elements 
3. Find length 
4. Merge 2 lists 
```

````python
list1 = [1,2,3,4,6]
print("initial list1: ", list1)
print("length: ",len(list1))
list1.append(5)
print("list1 after appending 5: ",list1)
print("length of list1 after appending 5: ", len(list1))
list1.remove(2)
print("list1 after removing 2: ",list1)
list1.remove(1)
print("list1 after removing 1: ",list1)
list2 = [11,12,13,14]
print("list1: ",list1)
print("list2: ",list2)
list1.extend(list2)
print("list1 + list2: ",list1)
````
