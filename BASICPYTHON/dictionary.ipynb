```python
# Creating a dictinary 

lst = [1,2,3,4,5,5,5,5,1,2,3,4,6,6,6,6,66,6]

mydict = {}

#for i in range(len(lst)):      "key":"value"
for num in lst: 
    if num in mydict:
        mydict[num] += 1 
    else:
        mydict[num] = 1

print(mydict)            



```

```python
# Given an array of integers arr, return true if the number of occurrences of each value 
# in the array is unique or false otherwise.
# arr = [1,2,2,1,1,3]


arr = [1,2,2,1,1,3]

def uniqueOccurrences(arr):

    mydict = {}
    for i in arr:
        if i in mydict:
            mydict[i] += 1
        else:
            mydict[i] = 1 
    return mydict 

aa = uniqueOccurrences(arr)         
print(aa)
# [3,2,1] True   {3:1,2:1,1:1}
# [3,3,2,1] False

x1 = aa.values()
x2 = set(x1)

print(x1)
print(x2)

print(len(x1) == len(x2))

```
