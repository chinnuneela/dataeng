```python
# Given an array of integers arr, return true if the number of occurrences of each value 
# in the array is unique or false otherwise.
# arr = [1,2,2,1,1,3]


arr = [1,2,2,1,1,3]

def uniqueOccurrences(arr):

    mydict = {}
    for i in arr:
        if i in mydict:
            mydict[i] += 1
        else:
            mydict[i] = 1 
    return mydict 

aa = uniqueOccurrences(arr)         
print(aa)
# [3,2,1] True   {3:1,2:1,1:1}
# [3,3,2,1] False

x1 = aa.values()
x2 = set(x1)

print(x1)
print(x2)

print(len(x1) == len(x2))

```

```python
word1 = 'abca'
word2 = 'bca'

dict1 = {}
dict2 = {}

for i in word1:
    if i in dict1:
        dict1[i] += 1
    else:
        dict1[i] = 1


for i in word2:
    if i in dict2:
        dict2[i] += 1
    else:
        dict2[i] = 1

print(f"values in dictionary1 : {dict1}")
print(f"values in dictionary2 : {dict2}")

print(dict1 == dict2)
```

Given an integer array nums, return an array answer such that answer[i] is equal to the product of all the elements of nums except nums[i].

The product of any prefix or suffix of nums is guaranteed to fit in a 32-bit integer.

You must write an algorithm that runs in O(n) time and without using the division operation.

 

Example 1:

Input: nums = [1,2,3,4]
Output: [24,12,8,6]
Example 2:

Input: nums = [-1,1,0,-3,3]
Output: [0,0,9,0,0]

```python
nums = [1,2,3,4]
product = 1
answer = {}

for i in nums:
    product = product * i

for i in range(len(nums)):
    if nums[i] != 0:
        answer[i] = product/nums[i]

print(f"product is {product}")
print(f"answer is {answer}")

```
