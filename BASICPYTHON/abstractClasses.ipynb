```python
# any service 
import boto3 

client = boto3.client('secretsmanager',"us-east-1")
```

```python
secret = client.get_secret_value(SecretId="creds")

credentials = secret['SecretString']

# it was giving a string , can be converted into dict

import json 

creds_dict = json.dumps(credentials) # converted to dict 
# '{"user":"Yash","password":"Welcome!"}' => string 
creds_dict = {"user":"Yash","password":"Welcome!"} # dict 

username = creds_dict["user"]
password = creds_dict["password"]
```
