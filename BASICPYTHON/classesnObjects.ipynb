**Inheritance**

```python
class Animal:

    def __init__(self,name):
        self.name = name 

    def eat(self):
        print(f"{self.name} is eating ") 

    def speak(self):
        print(f"{self.name} is speaking") 

    def long_teeth(self):
        pass

    def gives_milk(self):
        pass 
```

```python
class Tiger(Animal):
    def speak(self):
        print(f"{self.name} roars")

    def hunt(self):
        print(f"{self.name} is hunting in the jungle ")    
```

```python
class cow(Animal):
    def speak(self):
        print(f"{self.name} don't roar but different sound ")

    def give_milk(self):
        print(f"{self.name} gives milk ")    
```

```python
t = Tiger("Sher")
t.speak()
t.eat()
t.hunt()
```

```python
c = cow("Gauri")
c.speak()
c.eat()
c.give_milk()
```

## Super keyword

```python
class Animal:

    def __init__(self,name,species):
        self.name = name
        self.species = species
        print(f"Animal created : {self.name} , Species : {self.species} ")

# child 
class Tiger(Animal):
    def __init__(self,name,stripes):
        super().__init__(name,"Tiger")
        self.stripes = stripes



```

```python
t = Tiger("Sheru",100)
```

# Template design Pattern

```python
# Abstract Classes
from abc import ABC , abstractmethod 

class Animal(ABC):

    def __init__(self,name):
        self.name = name 

    @abstractmethod
    def speak(self):
        pass 

    def eat(self):
        print("f{self.name} is eating")  
```

```python
class Tiger(Animal):

    def roars(self):
        print("It roars")

    def speak(self):
       print("Yes I speaks")
```

```python
t = Tiger("sheru")
t.roars()
```
