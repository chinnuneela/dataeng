```python
def greet(greeting, *args , **kwargs):
    print(f"\n {greeting}")

    print("Extra Attributes ..")  # [a,b,b,....]
    for ch in args:
        print(f"{ch}") 

    print("Printing extra keyword arguments ") # {key : value , }
    for a,b in kwargs.items():
        print(f"a:{a} , b:{b}")

    print("----------------------------------------")    


greet("Hi")  

greet("Hi","Bye","hello")

greet("Hi",1,2,location=1,time=2)   # key : Value 

```
